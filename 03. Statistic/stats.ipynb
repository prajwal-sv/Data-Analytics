```python
# Discrete data
# 1. Binomial
# a. There are only two possibilities
# b. Probability of a single occurence is known

# I stay in Mumbai for 17 days. Prob of rains on any given day 
# is 0.7
# A. What is the prob of exactly 7 days of rain out of 17
binom.pmf(7,17,0.7)   # probability mass function
```

```python
# B. What is the prob of max 7 days of rain out of 17
binom.pmf(0,17,0.7) + binom.pmf(1,17,0.7) + binom.pmf(2,17,0.7)
binom.cdf(7,17,0.7)    # cumulative distr function
```

```python
# C. What is the prob of min 7 days of rain out of 17
1 - binom.cdf(6,17,0.7)
```

```python
# 2. Poisson
# Average number of accidents reported is 40.
# A. What is the prob of exactly 38 accidents
poisson.pmf(38,40)
```

```python
# B. What is the prob of max 38 accidents
poisson.cdf(38,40)
```

```python
# C. What is the prob of min 38 accidents
1 - poisson.cdf(37,40)
```
